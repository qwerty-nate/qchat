```python
import cirq
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
# ── Encoding library ──────────────────────────────────────────────────────────
CHAR_TO_VEC = {
    'NULL': '0000', ' ':    '00001',
    'e':    '00010', 't':    '00011', 'a':    '00100', 'o':    '00101',
    'i':    '00110', 'n':    '00111', 's':    '01000', 'h':    '01001',
    'r':    '01010', 'd':    '01011', 'l':    '01100', 'c':    '01101',
    'u':    '01110', 'm':    '01111', 'w':    '10000', 'f':    '10001',
    'g':    '10010', 'y':    '10011', 'p':    '10100', 'b':    '10101',
    'v':    '10111', 'k':    '11000', 'j':    '11001', 'x':    '11010',
    'q':    '11011', 'z':    '11100',
    '.':    '11101', ',':    '11110',
    '?':    '11111'
}
VEC_TO_CHAR = {v: k for k, v in CHAR_TO_VEC.items()}
```

```python
i = cirq.NamedQubit('i')
j = cirq.NamedQubit('j')
k = cirq.NamedQubit('k')
a = cirq.NamedQubit('a')
b = cirq.NamedQubit('b')
c = cirq.NamedQubit('c')
d = cirq.NamedQubit('d')
e = cirq.NamedQubit('e')


#print('Circuit:')
#print(circuit)
#print('Results:')
#print(result1)

#cirq.plot_state_histogram(result1)
#plt.show()
```

```python
word =[]
input("Question: ")
for repeat in range(7):
  circuit = cirq.Circuit(cirq.X(i), cirq.X(j), cirq.X(k), 
                       cirq.H(a), cirq.H(b), cirq.H(c), cirq.H(d), cirq.H(e),
                       cirq.measure(i, j, k, a, b, c ,d ,e))
  simulator = cirq.Simulator()
  result1 = simulator.run(circuit, repetitions=1)
  for key in result1.measurements:
    letter = result1.measurements[key].flatten()[3:]
    letter = ''.join(str(x) for x in letter)
    char = VEC_TO_CHAR.get(letter, ' ')
    word.append(char)
print(''.join(word))
```
